```python
#─────────────────────────────────────────────────────────────
# 0. 설정 : 파일명만 바꿔 주세요
#─────────────────────────────────────────────────────────────
INPUT_FILE  = "merged_result.csv"    # 원본 데이터 CSV
OUTPUT_FILE = "merged_result_0605.csv"  # 결과 저장 파일

#─────────────────────────────────────────────────────────────
# 1. 라이브러리
#─────────────────────────────────────────────────────────────
import pandas as pd
import numpy as np

#─────────────────────────────────────────────────────────────
# 2. 컬럼 이름 매핑  (🇰🇷 → 내부에서 쓸 짧은 이름)
#─────────────────────────────────────────────────────────────
COL_DATE   = "게시일_dt"   # 게시일
COL_CHAN   = "채널ID"      # 채널 ID
COL_VIEWS  = "조회수"       # 조회수

#─────────────────────────────────────────────────────────────
# 3. 데이터 로드 & 정렬
#─────────────────────────────────────────────────────────────
df = pd.read_csv(INPUT_FILE)

# (1) 날짜 컬럼을 datetime 으로 변환
df[COL_DATE] = pd.to_datetime(df[COL_DATE])

# (2) 채널+게시일 기준 정렬
df = df.sort_values([COL_CHAN, COL_DATE]).reset_index(drop=True)

#─────────────────────────────────────────────────────────────
# 4. 업로드 간격 계산
#─────────────────────────────────────────────────────────────
# 4-1. 직전 1개 영상과의 간격
df["prev_date"]      = df.groupby(COL_CHAN)[COL_DATE].shift(1)
df["upload_interval"] = (df[COL_DATE] - df["prev_date"]).dt.days

# 4-2. 평균 업로드 간격 (현재 영상 이전까지)
def expanding_mean_gap(series: pd.Series) -> pd.Series:
    gaps = series.diff().dt.days
    return gaps.expanding().mean().shift(1)

df["mean_upload_interval"] = (
    df.groupby(COL_CHAN)[COL_DATE]
      .transform(expanding_mean_gap)
)

# 4-3. NaN → 999 로 대체
df[["upload_interval", "mean_upload_interval"]] = (
    df[["upload_interval", "mean_upload_interval"]]
      .fillna(999)
      .astype(int)
)

#─────────────────────────────────────────────────────────────
# 5. 조회수(log1p) 기반 점수 맵 작성
#─────────────────────────────────────────────────────────────
df["view_log"] = np.log1p(df[COL_VIEWS])

# (1) upload_interval 점수
interval_score_map = (
    df.groupby("upload_interval")["view_log"]
      .mean()
      .to_dict()
)

# (2) mean_upload_interval 점수
mean_interval_score_map = (
    df.groupby("mean_upload_interval")["view_log"]
      .mean()
      .to_dict()
)

#─────────────────────────────────────────────────────────────
# 6. 점수 컬럼 추가 (.map)
#─────────────────────────────────────────────────────────────
df["upload_interval_score"] = df["upload_interval"].map(interval_score_map)
df["mean_interval_score"]   = df["mean_upload_interval"].map(mean_interval_score_map)

#─────────────────────────────────────────────────────────────
# 7. 정리 & 저장
#─────────────────────────────────────────────────────────────
df.drop(columns=["prev_date", "view_log"], inplace=True)
df.to_csv(OUTPUT_FILE, index=False)

print("✅ 완료! 결과 파일:", OUTPUT_FILE)
print("➕ 추가된 컬럼 →", ["upload_interval", "mean_upload_interval",
                         "upload_interval_score", "mean_interval_score"])
```
